# Сбор артефактов экспериментов
Упаковка выходов подключённых ноутбуков в один ZIP. Обучение не выполняется.


```python
from pathlib import Path
import zipfile
root = Path("/kaggle/input")
output = Path("/kaggle/working/collected_artifacts.zip")
files = [p for p in root.rglob("*") if p.is_file() and p.suffix in (".csv", ".json", ".parquet", ".txt", ".joblib")]
if not files:
    raise ValueError("Нет входных артефактов")
with zipfile.ZipFile(output, "w", compression=zipfile.ZIP_DEFLATED) as bundle:
    for path in sorted(files):
        bundle.write(path, path.relative_to(root))
print(len(files), output, output.stat().st_size, flush=True)
```
